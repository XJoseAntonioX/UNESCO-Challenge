# Google Fact Check Claim Search

Search claims checked by fact-checking organizations using the Fact Check Tools API.

```python
import os
import requests
from dotenv import load_dotenv

load_dotenv()
API_KEY = os.environ["FACTCHECK_API_KEY"]
url = "https://factchecktools.googleapis.com/v1alpha1/claims:search"
params = {
    "key": API_KEY,
    "query": "la tierra es plana",  # Replace with the claim to check
    "languageCode": "es",
    "pageSize": 10,
}

response = requests.get(url, params=params, timeout=30)
response.raise_for_status()
results = response.json().get("claims", [])

for item in results:
    review = item.get("claimReview", [{}])[0]
    print(f"Claim: {item.get('text', 'N/A')}")
    print(f"Rating: {review.get('textualRating', 'N/A')}")
    print(f"Source: {review.get('publisher', {}).get('name', 'N/A')}")
    print(f"URL: {review.get('url', 'N/A')}\n")
```
